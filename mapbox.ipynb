```python
import requests
import pandas as pd

def get_journey_info(origin, destination, api_key):
    directions_url = f"https://api.mapbox.com/directions/v5/mapbox/driving/{origin[1]},{origin[0]};{destination[1]},{destination[0]}?access_token={api_key}"
    response = requests.get(directions_url)
    journey_data = response.json()
    journey_time = journey_data['routes'][0]['duration']
    journey_countries = journey_data['routes'][0]['legs'][0]['steps']
    countries_passed = []
    for country in journey_countries:
        if 'country' in country['maneuver']['location']:
            countries_passed.append(country['maneuver']['location']['country'])
    return journey_time, countries_passed

def create_journey_database(origin, destination, countries_passed, journey_time):
    data = {'Origin': [origin],
            'Destination': [destination],
            'Countries Passed': [countries_passed],
            'Journey Time': [journey_time]}
    df = pd.DataFrame(data)
    return df

def main(origin, destination, api_key):
    journey_time, countries_passed = get_journey_info(origin, destination, api_key)
    journey_df = create_journey_database(origin, destination, countries_passed, journey_time)
    return journey_df

api_key = "API KEY"
origin = [40.7128, 74.0060] # New York, NY
destination = [34.0522, 118.2437] # Los Angeles, CA
journey_df = main(origin, destination, api_key)
print(journey_df)
```
