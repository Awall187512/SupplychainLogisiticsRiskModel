```python
# Read the CSV file
df = pd.read_csv('Turas_Training.csv')

# Remove extra whitespace
df['Country1'] = df['Country1'].str.strip()
df['Country2'] = df['Country2'].str.strip()
df['Country3'] = df['Country3'].str.strip()
df['Country4'] = df['Country4'].str.strip()

# Filter the data based on the input values
country_filter = (df['Country1'] == 'Bulgaria') & (df['Country2'] == 'Greece') & (df['Country3'] == 'Albania') & (df['Country4'] == 'Serbia')
filtered_data = df[country_filter]

# Print the filtered data
print(filtered_data)

# Sum the 'Total' column of the filtered data
total = filtered_data['Total'].sum()
```
